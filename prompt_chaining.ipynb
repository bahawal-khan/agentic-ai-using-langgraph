```python
from langgraph.graph import StateGraph,START,END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()

llm = ChatGroq(
    model = "openai/gpt-oss-20b",
    temperature = 0
)
```

```python
class State(TypedDict):
    topic: str
    explaination: str
    simple_explaination: str
    final_answere: str
```

```python
# Node 1
def explain_topic(state: State):

    prompt = f"""
    Explain the following topic in detail:

    {state["topic"]}
    """

    response = llm.invoke(prompt)

    return {
        "explaination": response.content
    }
```

```python
# Node 2
def simplify_explaination(state: State):

    prompt = f"""
    Take the following explanation and rewrite it
    in very simple language so that a beginner can understand it:

    {state["explaination"]}
    """

    response = llm.invoke(prompt)

    return {
        "simple_explaination": response.content
    }
```

```python

# Node 3
def final_answere(state: State):

    prompt = f"""
    Convert the following explanation into
    clear and concise bullet points:

    {state["simple_explaination"]}
    """

    response = llm.invoke(prompt)

    return {
        "final_answere": response.content
    }
```

```python
## create a state graph

graph = StateGraph(State)

## add nodes to graph

graph.add_node('explain_topic',explain_topic)

graph.add_node('simplify_explaination', simplify_explaination)


graph.add_node('final_answere', final_answere)


## add edges to the graph


graph.add_edge(START,'explain_topic')

graph.add_edge('explain_topic','simplify_explaination')

graph.add_edge('simplify_explaination','final_answere')

graph.add_edge('final_answere',END)

## know compile to malle it a executable graph 

workflow = graph.compile()
```

```python
final_output = workflow.invoke({
    'topic': 'The impact of AI in Pakistan'
})

print(final_output['simple_explaination'])
```
